```python
import boto3
import pandas as pd
import io

# --- Set up S3 client ---
s3 = boto3.client("s3")

# --- Define your bucket and file path ---
bucket = "stock-market-data20"
key = "processed/2025-10-13/AAPL_data.parquet"  # Change to any file you want to inspect

# --- Download the file into memory ---
obj = s3.get_object(Bucket=bucket, Key=key)

# --- Load the parquet file into a DataFrame ---
df = pd.read_parquet(io.BytesIO(obj["Body"].read()))

# --- Display info ---
print("✅ Columns found:", df.columns.tolist())
print("\nSample data:")
display(df.head())
```
